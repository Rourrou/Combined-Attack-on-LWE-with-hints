```python
def generate_se_sigma_approx_hint(m, n, bino, eta, sigma, num_hint):
    V = []
    L = []

    for i in range(num_hint):
        if bino:
            D_v = build_centered_binomial_law(eta)
            values, probabilities = zip(*D_v.items())
            v = np.array(np.random.choice(values, size=m+n, p=probabilities))
        else:
            v = np.array([random.randint(-1665, 1665) for _ in range(m+n)])
            # print(v)

        noisy = np.random.normal(0, sigma)
        l = dbdd.leak(v)+noisy
        V.append(v)
        L.append(l)
    print("approximate hints have generated")
    return V,L
```
